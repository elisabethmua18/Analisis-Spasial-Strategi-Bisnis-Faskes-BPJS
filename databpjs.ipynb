```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Memuat dataset
df = pd.read_csv('Data_Faskes_BPJS2019.csv')

# 2. Melihat informasi dasar dataframe
print("--- INFO DATASET ---")
print(df.info())

# 3. Menampilkan 5 baris pertama
print("\n--- 5 BARIS PERTAMA ---")
display(df.head())
```

```python
# 1. Cek jumlah missing values di setiap kolom
print("Jumlah Missing Values:")
print(df.isnull().sum())

# 2. Membersihkan duplikasi berdasarkan KodeFaskes (jika ada)
df = df.drop_duplicates(subset=['KodeFaskes'])

# 3. Memecah kolom 'LatLongFaskes' menjadi Latitude dan Longitude
# Asumsi format pemisah adalah koma (',')
if 'LatLongFaskes' in df.columns:
    # Menghapus spasi atau karakter tak perlu jika ada, lalu split
    lat_long_split = df['LatLongFaskes'].str.split(',', expand=True)
    
    if lat_long_split.shape[1] >= 2:
        df['Latitude'] = pd.to_numeric(lat_long_split[0].str.strip(), errors='coerce')
        df['Longitude'] = pd.to_numeric(lat_long_split[1].str.strip(), errors='coerce')
        print("\nBerhasil memecah LatLongFaskes menjadi Latitude & Longitude.")

# Periksa kembali 2 baris teratas setelah penambahan kolom koordinat
display(df[['NamaFaskes', 'LatLongFaskes', 'Latitude', 'Longitude']].head(2))
```

```python
# 1. Menghitung jumlah faskes berdasarkan TipeFaskes
print("\n--- DISTRIBUSI TIPE FASKES ---")
tipe_faskes_counts = df['TipeFaskes'].value_counts()
print(tipe_faskes_counts)

# 2. Menghitung 10 Kota/Kabupaten dengan faskes terbanyak
print("\n--- TOP 10 KOTA/KABUPATEN DENGAN FASKES TERBANYAK ---")
top_kotakab = df['KotaKab'].value_counts().head(10)
print(top_kotakab)
```

```python
# Pengaturan gaya visualisasi
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 5))

# Plot distribusi Tipe Faskes
sns.countplot(data=df, x='TipeFaskes', order=df['TipeFaskes'].value_counts().index, palette='viridis')
plt.title('Distribusi Jumlah Faskes Berdasarkan Tipe', fontsize=14, fontweight='bold')
plt.xlabel('Tipe Faskes', fontsize=12)
plt.ylabel('Jumlah', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Plot Top 10 Kota/Kabupaten dengan Faskes terbanyak
plt.figure(figsize=(12, 6))
sns.countplot(data=df, y='KotaKab', order=df['KotaKab'].value_counts().head(10).index, palette='crest')
plt.title('Top 10 Kota/Kabupaten dengan Faskes Terbanyak', fontsize=14, fontweight='bold')
plt.xlabel('Jumlah Faskes', fontsize=12)
plt.ylabel('Kota/Kabupaten', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# 1. Cek 5 baris pertama dari kolom LatLongFaskes untuk melihat bentuk aslinya
print("Isi asli kolom LatLongFaskes:")
print(df['LatLongFaskes'].head(5))

# 2. Cek apakah ada kolom lain yang mungkin berisi angka latitude/longitude terpisah
print("\nNama-nama kolom dalam dataframe:")
print(df.columns.tolist())
```

```python
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Ambil bagian teks setelah tanda '=' (mengambil bagian 'lat,long')
coords_extracted = df['LatLongFaskes'].astype(str).str.split('=').str[-1]

# 2. Pecah string 'lat,long' tersebut menjadi dua kolom terpisah
lat_long_split = coords_extracted.str.split(',', expand=True)

# 3. Konversi ke tipe data numerik (Latitude dan Longitude)
df['Latitude'] = pd.to_numeric(lat_long_split[0].str.strip(), errors='coerce')
df['Longitude'] = pd.to_numeric(lat_long_split[1].str.strip(), errors='coerce')

# 4. Bersihkan baris yang kosong/NaN
df_clean_coord = df.dropna(subset=['Latitude', 'Longitude']).copy()

print(f"Jumlah baris data valid dengan koordinat: {len(df_clean_coord)}")

# 5. Menjalankan K-Means Clustering
if len(df_clean_coord) > 0:
    X = df_clean_coord[['Latitude', 'Longitude']]
    k = 5  # Jumlah zona/klaster yang diinginkan

    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    df_clean_coord['Cluster'] = kmeans.fit_predict(X)

    # Mendapatkan titik pusat (centroid) klaster
    centroids = kmeans.cluster_centers_
    print("\nTitik Koordinat Pusat Klaster (Rekomendasi Lokasi Strategis/Hub):")
    for i, centroid in enumerate(centroids):
        print(f"Zona {i+1} -> Latitude: {centroid[0]:.4f}, Longitude: {centroid[1]:.4f}")

    # Visualisasi Peta Klaster Sederhana
    plt.figure(figsize=(10, 6))
    sns.scatterplot(
        data=df_clean_coord, 
        x='Longitude', 
        y='Latitude', 
        hue='Cluster', 
        palette='Set1', 
        alpha=0.6, 
        s=20
    )
    plt.scatter(
        centroids[:, 1], centroids[:, 0], 
        s=200, c='black', marker='X', label='Pusat Zona (Centroid)'
    )
    plt.title('Klaster Sebaran Faskes untuk Optimasi Wilayah Bisnis', fontsize=14, fontweight='bold')
    plt.xlabel('Longitude')
    plt.ylabel('Latitude')
    plt.legend()
    plt.show()
else:
    print("Data koordinat kosong.")
```

```python
# 1. Filter hanya koordinat yang berada di wilayah geografis Indonesia yang sah
# Indonesia berada di sekitar Lat: -11 hingga 6, Long: 95 hingga 141
df_indo_coord = df_clean_coord[
    (df_clean_coord['Latitude'] >= -11) & (df_clean_coord['Latitude'] <= 6) &
    (df_clean_coord['Longitude'] >= 95) & (df_clean_coord['Longitude'] <= 141)
].copy()

print(f"Jumlah data faskes valid di wilayah Indonesia: {len(df_indo_coord)}")

# 2. Jalankan ulang K-Means dengan data yang sudah bersih
X_clean = df_indo_coord[['Latitude', 'Longitude']]
k = 5

kmeans_clean = KMeans(n_clusters=k, random_state=42, n_init=10)
df_indo_coord['Cluster'] = kmeans_clean.fit_predict(X_clean)

# 3. Cetak titik pusat (centroid) yang masuk akal
centroids_clean = kmeans_clean.cluster_centers_
print("\nTitik Koordinat Pusat Klaster yang Valid (Rekomendasi Lokasi Strategis):")
for i, centroid in enumerate(centroids_clean):
    print(f"Zona {i+1} -> Latitude: {centroid[0]:.4f}, Longitude: {centroid[1]:.4f}")

# 4. Visualisasi ulang peta sebaran yang bersih
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df_indo_coord, 
    x='Longitude', 
    y='Latitude', 
    hue='Cluster', 
    palette='viridis', 
    alpha=0.5, 
    s=15
)
plt.scatter(
    centroids_clean[:, 1], centroids_clean[:, 0], 
    s=250, c='red', marker='X', label='Pusat Zona (Centroid)'
)
plt.title('Peta Sebaran Faskes Indonesia yang Bersih & Klaster Zonasi Bisnis', fontsize=14, fontweight='bold')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.legend()
plt.show()
```

```python
# Menghitung jumlah faskes di setiap klaster/zona
zone_counts = df_indo_coord['Cluster'].value_counts().reset_index()
zone_counts.columns = ['Zona', 'Jumlah_Faskes']
zone_counts = zone_counts.sort_values('Zona')

print("--- DISTRIBUSI JUMLAH FASKES PER ZONA BISNIS ---")
print(zone_counts)
```
